```python
#!/usr/bin/env python3
import numpy as np

# ===== USER SETTINGS =====
input_file = "Tau_V1130_no2.dat"   # Your original flux file
output_file = "Tau_V1130_no2.dat"  # Output magnitude file
zeropoint = 20.44  # Adjust as needed for your system
# =========================

# Read the data (assuming columns: Time Flux FluxError)
data = np.loadtxt(input_file, comments="#")
time = data[:, 0]
flux = data[:, 1]
flux_err = data[:, 2]

# We normalize the flux first to a reasonable scale
median_flux = np.median(flux)
norm_flux = flux / median_flux
norm_flux_err = flux_err / median_flux

# Convert flux to magnitude
mag = -2.5 * np.log10(flux) + zeropoint
mag_err = (2.5 / np.log(10)) * (norm_flux_err / norm_flux)

# Write out to new file with header
np.savetxt(output_file, np.column_stack([time, mag, mag_err]),
           fmt="%.8f %.5f %.5f", comments='#')

print(f"Converted flux to magnitude and saved to {output_file}")
```
